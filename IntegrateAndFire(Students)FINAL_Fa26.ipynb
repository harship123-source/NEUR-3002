```python
# Jupyter: enable plot rendering inside the notebook (this allows us to actually show the output of the code visually)
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

# --- Neuron Parameters (realistic, in mV and pA) ---
C = 200.0
V_th = -55 #annottated
V_reset = -65.0
V_rest = -70 #annottated
I_ext = 300.0
dt = 0.1


T = 100          # total simulation time in ms
dt = 0.1         # time step in ms
time = np.arange(0, T, dt)  # time vector

# --- Initialize voltage and spike times ---
V = np.full_like(time, V_rest)  # set all values to resting voltage
spike_times = []

# --- Simulation---
for i in range(1, len(time)):
    dV = (I_ext / C) * dt       # ΔV = (I / C) * dt, in mV
    V[i] = V[i-1] + dV

    if V[i] >= V_th:
        V[i-1] = 40.0           # mark spike as +40 mV (for visualization)
        V[i] = V_reset
        spike_times.append(time[i])

# --- Plotting ---
plt.figure(figsize=(10, 4))
plt.plot(time, V, label='fix me')
plt.axhline(V_th, color='r', linestyle='--', label='fix me')
plt.xlabel('Time (ms)') #what should the x be?
plt.ylabel('Voltage (mv)') #what should the y be?
plt.title('Integrate and Fire Model') #what is a good title?
plt.grid(True)
plt.show()

```

```python

plt.figure(figsize=(10, 2))
plt.eventplot(spike_times, lineoffsets=1, linelengths=0.5, colors='black')
plt.xlabel('Time (ms)')
plt.yticks([]) # Remove y-axis ticks as it's a single neuron
plt.ylabel('Neuron')
plt.title('Raster Plot of Neuron Spikes')
plt.xlim(0, T) # Set x-axis limit to the simulation time
plt.show()
```
